```python
sample_size = 100000
```

```python
exponential_data=np.random.exponential(
    scale=1.0,
    size=sample_size
)
```

```python
uniform_data = np.random.uniform(
    low=0.0,
    high=1.0,
    size=sample_size
)
```

```python
normal_data = np.random.normal(
    loc=0.0,
    scale=1.0,
    size=sample_size
)
```

```python
plt.figure(figsize=(10, 6))
```

```python
plt.hist(
    exponential_data,
    density=True,
    bins=50,
    histtype="step",
    linewidth=2,
    label="Exponential"
)
```

```python
plt.hist(
    uniform_data,
    density=True,
    bins=100,
    histtype="step",
    linewidth=2,
    label="Uniform"
)
```

```python
plt.hist(
    normal_data,
    density=True,
    bins=100,
    histtype="step",
    linewidth=2,
    label="Normal",
    color="green"
)
```

```python
plt.axis([-2.5, 2.5, 0, 1.1])
```

```python
plt.legend(loc="upper left")
plt.title("Random Distributions")
plt.xlabel("Value")
plt.ylabel("Density")
```

```python

plt.grid(True, linestyle="--", alpha=0.5)
```

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
```

```python
X, Y = np.meshgrid(x, y)
```

```python
Z = X**2 + Y**2
```

```python
fig = plt.figure(figsize=(10, 8))
ax = plt.axes(projection="3d")
surf = ax.plot_surface(X, Y, Z, cmap="viridis")
ax.set_title("3D Surface Plot of $Z = X^2 + Y^2$")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
fig.colorbar(surf, shrink=0.5, aspect=5, label="Z Value")
plt.show()
```

```python
df = pd.read_csv(
    'Pokemon.csv',
    index_col=0,
    encoding='latin'
)
```

```python
df.head()
```

```python
features = [
    "HP",
    "Attack",
    "Defense",
    "Sp. Atk",
    "Sp. Def",
    "Speed"
]
```

```python
selected_data = df[features]
```

```python
pearson_corr = selected_data.corr(method="pearson")
spearman_corr = selected_data.corr(method="spearman")
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    pearson_corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    square=True,
    linewidths=0.5
)

plt.title("Pearson Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    spearman_corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    square=True,
    linewidths=0.5
)

plt.title("Spearman Rank Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16,6))

sns.heatmap(
    pearson_corr,
    annot=True,
    cmap="coolwarm",
    ax=axes[0]
)
axes[0].set_title("Pearson")

sns.heatmap(
    spearman_corr,
    annot=True,
    cmap="coolwarm",
    ax=axes[1]
)
axes[1].set_title("Spearman")

plt.tight_layout()
plt.show()
```
